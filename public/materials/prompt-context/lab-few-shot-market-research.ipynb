```python
# ==========================================================
# TO DO 1. ChatOpenAI 객체 생성
# ==========================================================
#
# OpenAI GPT 모델을 사용할 수 있도록
# ChatOpenAI 객체를 생성하세요.
#
# 조건
#
# - Model : gpt-4.1-mini
# - Temperature : 0.3
#
# 생성한 객체를
#
# llm 변수에 저장하세요.
#
# ==========================================================


!pip install -q langchain-openai langchain-core

import os
from google.colab import userdata

OPENAI_API_KEY = userdata.get("OPENAI_API_KEY")
os.environ["OPENAI_API_KEY"] = OPENAI_API_KEY

from langchain_openai import ChatOpenAI
from langchain_core.prompts import (
    ChatPromptTemplate,
    FewShotChatMessagePromptTemplate
)
from langchain_core.output_parsers import StrOutputParser


llm = ChatOpenAI(
    model="gpt-4.1-mini",
    temperature=0.3
)

```

```python
# ==========================================================
# TO DO 2. 시장조사 Example 작성
# ==========================================================
#
# Few-shot Prompt에서 사용할 Example을 작성하세요.
#
# 다음 두 개의 산업을 Example로 작성합니다.
#
# Example 1
#
# 전기차 시장
#
# Example 2
#
# 스마트팩토리 시장
#
# 각 Example에는 반드시 다음 내용을 포함하세요.
#
# • 시장 규모
# • 주요 경쟁사
# • 시장 전망
#
# 작성한 Example은
#
# examples 변수에 저장하세요. 2 3 4 10 11
#
# ==========================================================


examples = [
    {
        "industry": "전기차 시장",
        "analysis": """
- 시장 규모: 2023년 글로벌 전기차 판매량은 약 1,400만 대 수준이며, 시장 규모는 지속적으로 확대되고 있다.
- 주요 경쟁사: 테슬라(Tesla), BYD, 폭스바겐(Volkswagen), 현대자동차·기아, GM 등이 주요 경쟁사이다.
- 시장 전망: 친환경 정책 강화, 배터리 기술 발전, 충전 인프라 확대로 전기차 시장은 장기적으로 성장할 전망이다.
""",
    },
    {
        "industry": "스마트팩토리 시장",
        "analysis": """
- 시장 규모: 글로벌 스마트팩토리 시장은 2023년 약 1,300억 달러 규모로 평가되며, 제조업 디지털 전환에 따라 성장하고 있다.
- 주요 경쟁사: 지멘스(Siemens), 로크웰 오토메이션(Rockwell Automation), 슈나이더 일렉트릭(Schneider Electric), ABB, 미쓰비시전기 등이 주요 경쟁사이다.
- 시장 전망: AI, IoT, 로봇, 디지털 트윈 도입 확대로 스마트팩토리 시장은 지속적으로 성장할 전망이다.
""",
    },
]
```

```python


# ==========================================================
# TO DO 3. Example Prompt 작성
# ==========================================================
#
# ChatPromptTemplate를 이용하여
# example_prompt를 작성하세요.
#
# Prompt는
#
# Human Message
#
# AI Message
#
# 순서로 구성합니다.
#
# 작성한 Prompt는
#
# example_prompt 변수에 저장하세요.
#
# ==========================================================

example_prompt = ChatPromptTemplate.from_messages([
    ("human", "산업 : {industry}"),
    ("ai", "{analysis}")
])


```

```python


# ==========================================================
# TO DO 4. Few-shot Prompt 생성
# ==========================================================
#
# FewShotChatMessagePromptTemplate를 이용하여
# Few-shot Prompt를 생성하세요.
#
# 다음 변수를 연결합니다.
#
# • examples
# • example_prompt
#
# 생성한 Prompt는
#
# few_shot_prompt 변수에 저장하세요.
#
# ==========================================================


few_shot_prompt = FewShotChatMessagePromptTemplate(
    example_prompt=example_prompt,
    examples=examples
)


```

```python

# ==========================================================
# TO DO 5. 최종 Prompt 작성
# ==========================================================
#
# ChatPromptTemplate를 이용하여
# 최종 Prompt를 작성하세요.
#
# 다음 순서로 구성합니다.
#
# ① System Prompt
#
# "당신은 Gartner의 Senior Research Analyst입니다."
#
# ② Few-shot Prompt
#
# ③ Human Prompt
#
# "{input}"
#
# 작성한 Prompt는
#
# final_prompt 변수에 저장하세요.
#
# ==========================================================


from langchain_core.prompts import ChatPromptTemplate

final_prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
당신은 반려동물 산업 전문 시장 분석 컨설턴트입니다.

시장조사 결과를 경영진 보고서 형태로 작성해야 합니다.
반드시 아래 구조를 유지하세요.

# 반려동물 헬스케어 시장 분석 보고서

## 1. 시장 개요
- 시장 현황
- 시장 규모 및 성장 방향

## 2. 시장 성장 요인
- 기술 요인
- 산업 요인
- 고객 요구 변화

## 3. 주요 고객 분석
- 주요 고객군
- 고객별 요구사항

## 4. 경쟁 환경
- 주요 경쟁 서비스 및 기업
- 경쟁 우위 요소

## 5. 사업 기회
- 신규 서비스 기회
- 차별화 전략

## 6. 위험 요소
- 시장 위험
- 기술 위험
- 사업 운영 위험

답변은 기업 의사결정자가 활용할 수 있도록 논리적이고 구체적으로 작성하세요.
"""
    ),few_shot_prompt,
    (
        "human",
        """
다음 시장에 대한 분석 보고서를 작성하세요.
한 발 물러서서 더 큰 그림을 봐주세요.
큰 맥락을 먼저 생각한 후에 보고서를 작성해주세요.
결론 전에 추론 과정을 먼저 보여주세요. 왜 그렇게 판단했는지 과정을 설명해주세요.

분석 대상 시장:
반려동물 헬스케어 시장

추가 분석 기준:
펫보험, 원격 건강관리, 웨어러블 기기, 맞춤형 영양 서비스를 중심으로 분석하세요.
"""
    )
])



```

```python


# ==========================================================
# TO DO 6. Prompt 실행
# ==========================================================
#
# 다음 질문을 실행하세요.
#
# "생성형 AI 교육 시장을 조사하세요."
#
# 실행 결과를 출력하세요.
#
# ==========================================================

from langchain_core.output_parsers import StrOutputParser


messages = final_prompt.format_messages(
    input="생성형 AI 교육 시장을 조사하세요."
)

response = llm.invoke(messages)

print(response.content)

```

```python

# ==========================================================
# TO DO 7. StrOutputParser 생성
# ==========================================================
#
# StrOutputParser 객체를 생성하세요.
#
# 생성한 객체를
#
# parser 변수에 저장하세요.
#
# ==========================================================



parser = StrOutputParser()
```

```python

# ==========================================================
# TO DO 8. LCEL Chain 구현
# ==========================================================
#
# LCEL을 이용하여
#
# Prompt
#
# ↓
#
# ChatOpenAI
#
# ↓
#
# StrOutputParser
#
# 순서의 Chain을 작성하세요.
#
# 생성한 Chain은
#
# chain 변수에 저장하세요.
#
# ==========================================================


chain = (
    final_prompt
    |
    llm
    |
    parser
)

```

```python
# ==========================================================
# TO DO 9. LCEL Chain 실행
# ==========================================================
#
# 작성한 Chain을 실행하세요.
#
# 질문
#
# "생성형 AI 교육 시장을 조사하세요."
#
# 실행 결과를 출력하세요.
#
# ==========================================================


result = chain.invoke(
    {
        "market":
        "스마트팩토리 시장"
    }
)

print("\n========== 결과 ==========\n")

print(result)
```

```python
# ==========================================================
# TO DO 10. 전략수립 Example 작성
# ==========================================================
#
# Few-shot Prompt에서 사용할
# 전략수립 Example을 작성하세요.
#
# Example 1
#
# 온라인 쇼핑몰
#
# Example 2
#
# AI 교육
#
# 각 Example에는
#
# • 단기 전략
# • 중기 전략
# • 장기 전략
#
# 을 포함합니다.
#
# ==========================================================

strategy_examples = [
    {
        "industry": "온라인 쇼핑몰",
        "strategy": """
- 단기 전략: 인기 상품 중심의 할인·쿠폰 프로모션을 운영하고, SNS 광고와 검색 광고를 통해 신규 고객 유입을 확대한다.
- 중기 전략: 고객 구매 데이터를 활용한 개인화 추천 기능과 멤버십 프로그램을 구축해 재구매율을 높인다.
- 장기 전략: 자체 브랜드(PB) 상품을 개발하고, 물류 자동화 및 해외 시장 진출을 통해 수익성을 강화한다.
"""
    },
    {
        "industry": "AI 교육",
        "strategy": """
- 단기 전략: 생성형 AI 기초와 실무 활용 중심의 온라인 강의를 개설하고, 기업 대상 단기 교육 프로그램을 운영한다.
- 중기 전략: 산업별 맞춤형 AI 교육 과정과 프로젝트 기반 실습 콘텐츠를 개발해 교육 품질을 차별화한다.
- 장기 전략: AI 교육 플랫폼을 구축하고, 대학·기업·공공기관과의 협력을 통해 국내외 교육 시장으로 확장한다.
"""
    }
]
```

```python

# ==========================================================
# TO DO 11. 전략수립 Few-shot Prompt 작성
# ==========================================================
#
# 전략수립 Example을 이용하여
#
# FewShotChatMessagePromptTemplate를 작성하세요.
#
# 생성한 Prompt를
#
# strategy_few_shot 변수에 저장하세요.
#
# ==========================================================


from langchain_core.prompts import (
    ChatPromptTemplate,
    FewShotChatMessagePromptTemplate
)

strategy_example_prompt = ChatPromptTemplate.from_messages([
    ("human", "{industry}"),
    ("ai", "{strategy}")
])

strategy_few_shot = FewShotChatMessagePromptTemplate(
    examples=strategy_examples,
    example_prompt=strategy_example_prompt
)
```

```python

# ==========================================================
# TO DO 12. 전략수립 Prompt 작성
# ==========================================================
#
# 다음 System Prompt를 작성하세요.
#
# "당신은 McKinsey의 전략 컨설턴트입니다."
#
# Few-shot Prompt와 Human Prompt를 연결하여
#
# 최종 Prompt를 작성하세요.
#
# ==========================================================

strategy_prompt = ChatPromptTemplate.from_messages([
    ("system", """

    당신은 McKinsey의 전략 컨설턴트입니다.

    시장 진출 전략이 실패했다고 가정하세요.
    왜 전략이 실패했을 지 분석해주세요.
    그리고 실패원인을 바탕으로 더 구체적이고 전략적으로 보고서를 작성하세요.
    """),
    strategy_few_shot,
    ("human", "{industry} 사업의 전략을 수립해줘.")
])
```

```python
# ==========================================================
# TO DO 13. LCEL Chain 구현
# ==========================================================
#
# 전략수립 Prompt를 이용하여
#
# LCEL Chain을 작성하세요.
#
# Prompt
#
# ↓
#
# ChatOpenAI
#
# ↓
#
# StrOutputParser
#
# ==========================================================


strategy_chain = strategy_prompt | llm | parser


```

```python

# ==========================================================
# TO DO 14. 전략수립 실행
# ==========================================================
#
# 다음 질문을 실행하세요.
#
# "생성형 AI 교육 서비스의 전략을 수립하세요."
#
# 실행 결과를 출력하세요.
#
# ==========================================================

result = strategy_chain.invoke({
    "industry": "생성형 AI 교육 서비스의 전략을 수립하세요."
})

print(result)
```

```python


# ==========================================================
# TO DO 15. 의사결정 Example 작성
# ==========================================================
#
# Few-shot Prompt에서 사용할
# 의사결정 Example을 작성하세요.
#
# Example
#
# 전기차 시장 진출
#
# 다음 내용을 포함합니다.
#
# • 진출 추천
# • 시장 성장률
# • 정부 지원 정책
# • 수익성
#
# ==========================================================

decision_examples = [
    {
        "topic": "전기차 시장 진출",
        "decision": """
- 진출 추천: 조건부 진출을 추천합니다. 배터리·충전 인프라·소프트웨어 등 차별화 가능한 분야에 집중해야 합니다.
- 시장 성장률: 글로벌 전기차 시장은 중장기적으로 높은 성장세가 예상되지만, 단기적으로는 보조금 축소와 가격 경쟁의 영향을 받을 수 있습니다.
- 정부 지원 정책: 주요 국가들은 친환경차 보급 확대, 세제 혜택, 충전 인프라 투자 등 지원 정책을 시행하고 있습니다.
- 수익성: 완성차 시장은 경쟁이 치열해 수익성 확보가 어렵지만, 배터리·부품·충전·서비스 분야는 성장성과 수익 기회가 상대적으로 높습니다.
"""
    }
]
```

```python

# ==========================================================
# TO DO 16. 의사결정 Few-shot Prompt 작성
# ==========================================================
#
# 의사결정 Example을 이용하여
#
# FewShotChatMessagePromptTemplate를 작성하세요.
#
# ==========================================================

decision_example_prompt = ChatPromptTemplate.from_messages([
    ("human", "{topic}에 대한 의사결정을 내려줘."),
    ("ai", "{decision}")
])

decision_few_shot = FewShotChatMessagePromptTemplate(
    examples=decision_examples,
    example_prompt=decision_example_prompt
)
```

```python


# ==========================================================
# TO DO 17. System Prompt 작성
# ==========================================================
#
# 다음 System Prompt를 작성하세요.
#
# "당신은 BCG의 Senior Partner입니다."
#
# ==========================================================

decision_system_prompt = """
당신은 AI 기반 업무 자동화 프로젝트를 평가하는
CEO 와 함께 심의에 참석한 최종 프로젝트 분석 전문가입니다.

지금까지 수행한 시장조사,
사업 전략 수립,
의사결정 결과를 종합하여
최종 프로젝트 보고서를 작성하세요.

분석 대상 사업은 국내 드론 배송 서비스 시장입니다.

드론 배송 서비스는 무인비행장치를 활용하여
음식, 생활용품, 의약품, 긴급 물품 등을
고객에게 배송하는 물류 서비스입니다.

반드시 다음 내용을 포함하세요.

# AI 기반 드론 배송 서비스 시장 분석 및 사업 전략 프로젝트 결과 보고서

## 1. 프로젝트 개요

## 2. 최종 시장 분석 결과

## 3. 사업 전략 결과

## 4. 최종 의사결정

## 5. 실무 활용 방안

## 6. 개선 방향


보고서는 실제 기업 경영진에게
제출 가능한 수준으로 작성하세요.
한발 뒤로 물러서서 큰 그림을 먼저 보고, 생각해주세요.
1. AI 기반 드론 배송 서비스 시장에서 겪는 근본적인 문제는 무엇일지
2. 기존 시장의 문제점을 해결하지 못한 이유가 무엇인가?
3. 본 시장 연구가 기여할 수 있는 새로운 연구 방향을 제안하라

확인되지 않은 비용, 매출, 시장 규모는
사실처럼 단정하지 말고 추가 검증이 필요하다고 표현하세요.
"""

```

```python


# ==========================================================
# TO DO 18. 최종 Prompt 작성
# ==========================================================
#
# 다음 순서로 Prompt를 작성하세요.
#
# ① System Prompt
#
# ② Few-shot Prompt
#
# ③ Human Prompt
#
# 작성한 Prompt를
#
# decision_prompt 변수에 저장하세요.
#
# ==========================================================
decision_prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system", decision_system_prompt
        ),
decision_few_shot,
        (
            "human",
            """
다음 프로젝트 결과를 기반으로
최종 보고서를 작성하세요.

[시장조사 결과]

{market_result}

[사업 전략 결과]

{strategy_result}

[의사결정 결과]

{decision_result}

[평가 결과]

{evaluation_result}
"""
        )
    ]
)
```

```python
# ==========================================================
# TO DO 19. LCEL Chain 구현
# ==========================================================
#
# decision_prompt를 이용하여
#
# LCEL Chain을 작성하세요.
#
# decision_prompt
#
# ↓
#
# ChatOpenAI
#
# ↓
#
# StrOutputParser
#
# ==========================================================


final_report_chain = (
    decision_prompt
    |
    llm
    |
    parser
)
```
